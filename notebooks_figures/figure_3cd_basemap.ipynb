# Making a basemap

**Purpose**: Create a nice basemap to use for Figure 3 C and D results (which are overlaid on top of it)

# Imports

```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import geopandas as gpd
import matplotlib.pyplot as plt
from pyproj import Transformer
```

### Manually set map boundaries

```python
xmin,ymin,xmax,ymax = [-130.,24.,-58.134777270438285,54.95639625203041]
```

### Plot

```python
# specify a mercator map
fig = plt.figure(figsize=(12,7))
ax = plt.axes(projection=ccrs.Mercator()) # this is epsg3857 = web-mercator, so we can get a scalebar in m


ax.set_extent([xmin, xmax, ymin, ymax], crs=ccrs.PlateCarree()) 

# add ocean and land 
# more here: https://scitools.org.uk/cartopy/docs/latest/matplotlib/feature_interface.html
ax.add_feature(cfeature.OCEAN, facecolor="grey")
ax.add_feature(cfeature.LAND,  facecolor="white")

# put state boundaries on top
ax.add_feature(cfeature.STATES, edgecolor="gray", linewidth=0.1)

# add our vertical line to separate sides
ax.plot(
    [-120, -120], [ymin, ymax],
    color="black", linestyle="--", linewidth=1,
    transform=ccrs.PlateCarree()
)

ax.plot(
    [-70, -70], [ymin, ymax],
    color="black", linestyle="--", linewidth=1,
    transform=ccrs.PlateCarree()
)

ax.plot(
    [-100, -100], [ymin, ymax],
    color="black", linestyle="--", linewidth=1,
    transform=ccrs.PlateCarree()
)

ax.plot(
    [-97.668406996997, -97.668406996997], [ymin, ymax],
    color="red", linestyle="--", linewidth=1,
    transform=ccrs.PlateCarree()
)

ax.axis("off")

plt.tight_layout()
plt.savefig("../figures/figures_current/components/figure_3cd_basemap.pdf")
plt.show()
```
